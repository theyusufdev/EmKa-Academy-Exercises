```python
# 2) donem1_mat adında bir seri oluşturun. Bu serinin içinde "Ali" ,"Deniz" ,"Aslı" ,"Burak" ,"Kerem" ,"Yasemin" ,"Derya" index değerleri olsun, 
# 85, 80 , 94 , 74 , 91 , 96 , 70 data değerleri olsun. Bu şekilde donem1 adında öğrencilerin 1.dönem matematik nor ortalamalarının tutulduğu bir seri oluşturun.
# donem2_mat adında başka bir seri oluşturun. Okula yeni bir öğrenci gelmiş olsun ve indexlerde bu sefer "Ali" ,"Deniz" ,"Aslı" ,"Burak" ,"Kerem" ,"Yasemin" ,"Derya" ,"Mert"         
# indexleri olsun. 80, 92 , 85 , 87, 90 , 92, 100, 83 ise dataları olsun. Bu iki seriyi birleştirerek bütün öğrencilerin yıl sonu matematik ORTALAMASINI hesaplayın.
# Not : Eğer bir öğrencinin sadece ikinci dönem notu bulunuyorsa, birinci dönem hesaba eklenmeden ikinci dönem notu yıl sonu notu olarak hesaplansın.

scores_1 = {"Ali": 85, "Deniz": 80, "Aslı": 94, "Burak": 74, "Kerem": 91, "Yasemin": 96, "Derya": 70}
scores_2 = {"Ali": 80, "Deniz": 92, "Aslı": 85, "Burak": 87, "Kerem": 90, "Yasemin": 92, "Derya": 100, "Mert": 83}

donem1_mat = pd.Series(data=scores_1)
donem2_mat = pd.Series(data=scores_2)

# Bu kod tanımlı her bir öğrencinin notu varmı diye kontrol eder varsa True, yoksa False döndürür
# ama .astype(int) sayesinde True=1, False=0 olacak şekilde sayı döndürür bunları her bir döenm için yapar
# ve iki dönem notu olanın dönem sayısı 2, bir dönem notu olanaın dönem sayısı 1 olur
donem_sayilari = donem1_mat.notna().astype(int).add(donem2_mat.notna().astype(int), fill_value=0)

total_donem = (donem1_mat.add(donem2_mat, fill_value=0)) / donem_sayilari
total_donem
```
